### File I/O 
python can be used to perform operations on a file .(read & write data)
### Types of all files  
1 : Text files :where data is stored in characters e.g; txt , docx , .log etc
2 : Binary files :where data is stored other than characters e.g; .mp4 , .mov , .png , .jpeg etc.

```python
# FILE HANDLING:
s=open ('text.txt' , 'r')
print(s)
```

```python
s=open ('text.txt' , 'r')          # READING A FILE:
text=s.read()
print(text)
s.close()
```

```python
s=open ('text.txt' , 'r')          # READING A FILE:
text=s.read(5)
print(text)
s.close()
```

```python
s=open ('text.txt' , 'r')          # READING A FILE LINE BY LINE 
text=s.readline()
print(text)
s.close()
```

```python
s=open ('text.txt' , 'w')    # WRITING A FILE: rewrites the whole data with new data
text=s.write('this is what im writing in write mode')
print(text)
s.close()
```

```python
s=open ('text.txt' , 'a')    # APPEND: adds the new data at the end of the existing data
text=s.write('\nThis is new data which i appends')
print(text)
s.close()
```

```python
# r+ = Read and Write (without truncating the file)

s=open ('text.txt' , 'r+') 
text=s.read()
print(text)
s.close()
```

```python
# r+ = Read and Write (without truncating the file)

s=open ('text.txt' , 'r+')        # write in here  overwrites characters, not lines.
text=s.write('mehreen')
print(text)
s.close()
```

```python
s=open ('text.txt' , 'r+')       
text=s.write('mehreen')
print(text)
print(s.read())
s.close()
```

```python
# w+ = Write and Read (truncate existing file) 
# If the file already has data, it will be erased immediately when you open it in w+ mode.

s=open ('text.txt' , 'w+') 
print(s.read())    
print(s.write('hey ! I am back'))
s.close()
```

```python
# With syntax : When using with syntax its not necessary to close the file because it does it automatically . 

with open ('text.txt' , 'r') as f:
    data = f.read()
data
```

```python
# Deleting a file : using os module 
# Deleted file cannot be recovered
# File must be closed before deleting
# Raises FileNotFoundError if file does not exist

import os
os.remove('text.txt')
```

```python
# Creating file : if we open file in write mode and file does not exists it automatically creates it . 

with open ('practice.txt','w') as p:
    data = p.write ('hi everyone \n we are learning file I/O \n using java\n like programming in java')
print(data)
```

```python
# WAF that replaces all occurrences of 'java' with 'python' in above file . 
with open ('practice.txt','r') as p:
    data1 = p.read()
data1
new_data = data1.replace('java','python')
new_data
    
```

```python
with open ('practice.txt','w') as p:
    data2 = p.write(new_data)
```
