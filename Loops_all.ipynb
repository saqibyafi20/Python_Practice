# For Loop 
- A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

#### Range function
- To loop through a set of code a specified number of times, we can use the range() function,
- The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

### ELSE Keyword in loop
- Suppose you allow a user to try 3 attempts to enter the correct password.
- If they succeed, break the loop.
- If all 3 attempts fail, show a message like "Account Locked."

- In Python, the else block with a for loop runs only if the for loop finishes normally — meaning, it did not break.

- If the loop is stopped using break, the else block will not execute.

## break
- Used to terminate the loop

```python
nums = [22,11,55,44,33,66,88,77,10,20,30,40,50,60,70,80,90,15,25,35,45,55,65,75,85,95]
x = int(input('Enter the number : '))

count = 0
for num in nums:
    count += 1
    if num == x:
        print(x, 'is present in the list')
        
print('Number of iterations : ', count)
```

```python
nums = [22,11,55,44,33,66,88,77,10,20,30,40,50,60,70,80,90,15,25,35,45,55,65,75,85,95]
x = int(input('Enter the number : '))

count = 0
for num in nums:
    count += 1
    if num == x:
        print(x, 'is present in the list')
        break
        
print('Number of iterations : ', count)
```

```python
# Given a list of numbers take a number as input from the user and check whether the number is present in the list or not
nums = [22,11,55,44,33,66,88,77,10,20,30,40,50,60,70,80,90,15,25,35,45,55,65,75,85,95]

x = int(input('Enter the number : '))

flag = 0
for num in nums:
    if num == x:
        flag = 1
        break

#print(flag)
if flag == 1:
    print(x,'is present in the list')
else:
    print(x,'is not present in the list')
```

## Continue 
- With the continue statement we can stop the current iteration of the loop, and continue with the next:

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

```python
for num in range(1, 6):
    if num == 3:
        print("Skipping", num)
        continue
    print("Number is:", num)
```

```python
for num in range(1, 10):
    if num == 7:
        print("Breaking at 7")
        break
    if num % 2 == 0:
        continue
    print(f"Number is {num}")

```

## for else loop in python
- The else keyword in a for loop specifies a block of code to be executed when the loop is finished:



```python
nums = [22,11,55,44,33,66,88,77,10,20,30,40,50,60,70,80,90,15,25,35,45,55,65,75,85,95]
x = int(input('Enter the number : '))

for num in nums:
    if num == x:
        print(x, 'is present in the list')
        break
else:
    print(x,'is not present in the list')
```

```python
nums = [11,55,22,77,44,66,88,91,31,46,53,50]

x = int(input('Enter the number : '))
i = 0
while i < len(nums):
    if nums[i] == x:
        print('The number {} is present in the list'.format(x))
        break
    i += 1
else:
    print('The number {} is not present in the list'.format(x))
```

```python
# given a list of numbers find the frequency of occurrence of distinct numbers
nums = [11,22,33,11,33,22,11,11,33,11]

freq = {}
for num in nums:
    if num in freq.keys():
        freq[num] += 1
    else:
        freq[num] = 1
        
print(freq)
```

```python
# create a program that takes a number as input and prints whether it is prime or not
num = int(input('Enter the number : '))

if num < 2:
    print(num,'is not a prime number')
else:
    for i in range(2, (num // 2) + 1):
        if num % i == 0:
            print(num, 'is not a prime number')
            break
    else:
        print(num, 'is a prime number')
```

```python
# create a program that takes a number as input and prints whether it is prime or not
num = int(input('Enter the number : '))

if num < 2:
    print(num,'is not a prime number')
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, 'is not a prime number')
            break
    else:
        print(num, 'is a prime number')
```

```python
num = int(input('Enter the number : '))

if num < 2:
    print(num,'is not a prime number')
else:
    for i in range(2, int(num ** (1/2)) + 1):
        if num % i == 0:
            print(num, 'is not a prime number')
            break
    else:
        print(num, 'is a prime number')
```
