```python
# MAP: the map function applies a function to each element in a sequence and returns a new sequence containing the transformed elements.syntax;
# map(function , iterable)
```

```python
# return cube of a list as list:

l=[2,4,6,8]
cube=[]
for i in l:
    cube.append(i**3)
print(cube)
```

```python
# return cube of a list as list: by map
l=[1,3,5,7,9]
cube=lambda x: x*x*x
    
```

```python
new_lst=list(map(cube , l))
print(new_lst)
```

```python
l=[1,3,5,7,9]
new_lst=list(map(lambda x :x*x*x , l))   # we also can take direct lambda function as function.
print(new_lst)
```

```python
# FILTER : the filter function filters a sequence of elements based on a given prediction (a functions that returns a boolean value) and returns a 
# new sequence containing only the elements that meets the predicate.

def even_odd(a):
    if a%2==0:
         return True
    else:
        False

```

```python
l=[2,4,3,6,8,5]
# return a list with even numbers;

a=list(filter(even_odd,l))
print(a)
```

```python
lst=['12','15','20','23']
lst_new = list(map(int , lst))
print(lst_new)
```

```python
lst = ['apple','banana','cherry']
lst_upper = list(map(str.upper , lst))
print(lst_upper)
```

```python
lst=[0 , 20 , 37 , 100]
lst_new = list(map(lambda x : (x*9/5)+32 ,lst))
print(lst_new)
```

```python
lst1=[2,4,6]
lst2=[3,5,7]
lst3 = list(map(lambda x,y: x+y , lst1,lst2))
print(lst3)
```

```python
lst = ['cat','dog','mouse']
dic = dict(map(lambda x:(x,len(x)) , lst))
print(dic)
```

```python
old_lst = ['John doe' , 'jane smith' , 'PETER PAN']
new_lst = list(map(str.capitalize , old_lst))
print(new_lst)
```
